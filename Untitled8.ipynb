```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

def r_squared(y, y_hat):
    ss_res = np.sum((y-y_hat)**2)
    ss_tot = np.sum((y-np.mean(y))**2)
    return 1-ss_res/ss_tot

# --- Data ---
x = np.linspace(1, 10, 20)
y = 2.5*x + 1.8 + np.random.normal(0, 5, 20)

# --- Linear Fit: Normal Equations ---
n = len(x)
a = (n*np.sum(x*y)-np.sum(x)*np.sum(y))/(n*np.sum(x**2)-np.sum(x)**2)
b = (np.sum(y)-a*np.sum(x))/n

y_lin = a*x+b

print("Linear: y = {:.4f}x + {:.4f}".format(a,b))
print("R² =", r_squared(y,y_lin))

# --- Visualization ---
xf = np.linspace(x.min(),x.max(),200)

plt.scatter(x,y,color='blue',label='Noisy Data')
plt.plot(xf,a*xf+b,'r-',lw=2,label='Linear Fit')

plt.title('Linear Fit - Noise = 5')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

def r_squared(y, y_hat):
    ss_res = np.sum((y-y_hat)**2)
    ss_tot = np.sum((y-np.mean(y))**2)
    return 1-ss_res/ss_tot

# --- Data ---
x = np.linspace(0,3,20)
y = 2.0*np.exp(-0.8*x) + np.random.normal(0,0.1,20)

# --- Exponential Fit ---
log_y = np.log(y)
ce = np.polyfit(x,log_y,1)

a_e = np.exp(ce[1])
b_e = ce[0]

y_exp = a_e*np.exp(b_e*x)

print("Exponential: y = {:.4f}*e^({:.4f}x)".format(a_e,b_e))
print("R² =",r_squared(y,y_exp))

# --- Visualization ---
xf = np.linspace(x.min(),x.max(),200)

plt.scatter(x,y,color='blue',label='Data')
plt.plot(xf,a_e*np.exp(b_e*xf),'r-',lw=2,label='Exponential Decay Fit')

plt.title('Exponential Curve Fitting - Decay')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

def r_squared(y, y_hat):
    ss_res = np.sum((y-y_hat)**2)
    ss_tot = np.sum((y-np.mean(y))**2)
    return 1-ss_res/ss_tot

# --- Generate Data ---
x = np.linspace(1,10,50)
y = 1.5*x**2 - 2*x + 3 + np.random.normal(0,3,50)

# --- Split Training and Testing Data ---
indices = np.random.permutation(len(x))

train_idx = indices[:40]
test_idx = indices[40:]

x_train = x[train_idx]
y_train = y[train_idx]

x_test = x[test_idx]
y_test = y[test_idx]

# --- Quadratic Fit Using Training Data ---
c = np.polyfit(x_train,y_train,2)

y_train_hat = np.polyval(c,x_train)
y_test_hat = np.polyval(c,x_test)

# --- Calculate R-squared ---
r2_train = r_squared(y_train,y_train_hat)
r2_test = r_squared(y_test,y_test_hat)

# --- Calculate RMSE ---
rmse_train = np.sqrt(np.mean((y_train-y_train_hat)**2))
rmse_test = np.sqrt(np.mean((y_test-y_test_hat)**2))

print("Quadratic coefficients:",c)
print("Training R² =",r2_train)
print("Testing R² =",r2_test)
print("Training RMSE =",rmse_train)
print("Testing RMSE =",rmse_test)

# --- Visualization ---
xf = np.linspace(x.min(),x.max(),200)
yf = np.polyval(c,xf)

plt.scatter(x_train,y_train,color='blue',label='Training Data')
plt.scatter(x_test,y_test,color='orange',label='Testing Data')
plt.plot(xf,yf,'r-',lw=2,label='Fitted Quadratic Curve')

plt.title('Training and Testing Dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid()
plt.show()
```
